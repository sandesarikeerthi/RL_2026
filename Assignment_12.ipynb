```python

# Value Iteration Algorithm

# Define states and actions
states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

# Discount factor
gamma = 0.9

# Initialize values
V = {'S0': 0, 'S1': 0, 'G': 0}

# Define transitions: (next_state, reward)
transitions = {
    'S0': {
        'Up': ('S0', -1),
        'Down': ('S0', -1),
        'Left': ('S0', -1),
        'Right': ('S1', -1)
    },
    'S1': {
        'Up': ('S1', -1),
        'Down': ('S1', -1),
        'Left': ('S1', -1),
        'Right': ('G', 100)
    }
}

# Value Iteration
theta = 0.001

while True:
    delta = 0
    new_V = V.copy()

    for s in states:
        if s == 'G':
            continue

        values = []

        for a in actions:
            next_state, reward = transitions[s][a]

            value = reward + gamma * V[next_state]
            values.append(value)

        new_V[s] = max(values)

        delta = max(delta, abs(new_V[s] - V[s]))

    V = new_V

    if delta < theta:
        break

# Display results
print("Optimal State Values:")

for s in states:
    print(s, "=", V[s])

print("\nOptimal Path: S0 -> S1 -> G")
```
