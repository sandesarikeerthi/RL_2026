```python
import pandas as pd

# Assignment 10: Bellman Optimality and Value Iteration
# Course Code: 23CS201SE403

# 1. Define the MDP

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

gamma = 0.9
theta = 0.001

normal_reward = -1
wall_reward = -1
goal_reward = 10

print("1. MDP INFORMATION")
print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)
print("Normal Reward:", normal_reward)
print("Wall Reward:", wall_reward)
print("Goal Reward:", goal_reward)
print("Terminal State: G")

print("\nGrid World:")
print("S0 | S1")
print("Wall | G")

# 2. Define transitions

transitions = {
    'S0': {
        'Up': ('S0', wall_reward),
        'Down': ('S0', wall_reward),
        'Left': ('S0', wall_reward),
        'Right': ('S1', normal_reward)
    },
    'S1': {
        'Up': ('S1', wall_reward),
        'Down': ('G', goal_reward),
        'Left': ('S0', normal_reward),
        'Right': ('S1', wall_reward)
    }
}

# 3. Verify transitions

print("\n2. TRANSITION VERIFICATION")

for state in ['S0', 'S1']:
    for action in actions:
        next_state, reward = transitions[state][action]
        print(
            f"State: {state}, Action: {action}, "
            f"Next State: {next_state}, Reward: {reward}"
        )
        assert next_state in states

print("All state-action pairs have defined transitions.")

# 4. Initialize state values

V = {'S0': 0.0, 'S1': 0.0, 'G': 0.0}

print("\n3. INITIAL STATE VALUES")
print(V)

# 5. Implement synchronous Value Iteration

print("\n4. VALUE ITERATION")

iteration = 0
history = []

while True:
    iteration += 1

    old_V = V.copy()
    new_V = old_V.copy()

    for state in ['S0', 'S1']:
        q_values = []

        for action in actions:
            next_state, reward = transitions[state][action]
            q = reward + gamma * old_V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    new_V['G'] = 0.0

    delta = max(
        abs(new_V[state] - old_V[state])
        for state in states
    )

    V = new_V

    history.append({
        'Iteration': iteration,
        'V(S0)': V['S0'],
        'V(S1)': V['S1'],
        'V(G)': V['G'],
        'Delta': delta
    })

    print(
        f"Iteration {iteration}: "
        f"V(S0)={V['S0']:.4f}, "
        f"V(S1)={V['S1']:.4f}, "
        f"V(G)={V['G']:.4f}, "
        f"Delta={delta:.4f}"
    )

    if delta < theta:
        break

# 6. Display iteration-wise values

print("\n5. ITERATION-WISE STATE VALUE TABLE")
df_history = pd.DataFrame(history)
print(df_history.to_string(index=False))

# 7. Display final optimal state values

print("\n6. FINAL OPTIMAL STATE VALUES")

for state in states:
    print(f"V*({state}) = {V[state]:.4f}")

# 8. Calculate action values

Q = {}

for state in ['S0', 'S1']:
    Q[state] = {}

    for action in actions:
        next_state, reward = transitions[state][action]
        Q[state][action] = reward + gamma * V[next_state]

print("\n7. ACTION VALUES Q*(s, a)")
df_Q = pd.DataFrame.from_dict(Q, orient='index')
print(df_Q.round(4).to_string())

# 9. Extract optimal policy

print("\n8. OPTIMAL POLICY")

policy = {}

for state in ['S0', 'S1']:
    max_q = max(Q[state].values())

    best_actions = [
        action for action in actions
        if abs(Q[state][action] - max_q) < 1e-9
    ]

    policy[state] = best_actions

    print(
        f"State {state}: Optimal Action(s) = {best_actions}, "
        f"Value = {max_q:.4f}"
    )

# 10. Bellman Optimality verification

print("\n9. BELLMAN OPTIMALITY VERIFICATION")

for state in ['S0', 'S1']:
    max_q = max(Q[state].values())
    verified = abs(V[state] - max_q) < 1e-9

    print(
        f"State: {state}, "
        f"V*(s) = {V[state]:.4f}, "
        f"max Q*(s,a) = {max_q:.4f}, "
        f"Verified = {verified}"
    )

    assert verified

print("\nAll Bellman Optimality checks passed.")

# 11. Result and observations

print("\n10. RESULT AND OBSERVATIONS")
print("1. Value Iteration calculates optimal state values.")
print("2. The goal provides a reward of +10.")
print("3. Normal moves and wall collisions give a reward of -1.")
print("4. Gamma = 0.9 discounts future rewards.")
print("5. Synchronous updates use previous iteration values.")
print("6. Iteration stops when Delta is smaller than theta.")
print("7. The optimal policy selects actions with maximum Q-values.")
print("8. Bellman Optimality is verified for all non-terminal states.")

print("\nAssignment completed successfully.")
```
